# Asteroid Hazard Classification — Exploratory Analysis
This notebook explores the dataset of 4,534 asteroids with 24 features to understand data quality, distributions, and relationships related to the target variable `Hazardous`.

```python
# Imports and dataset loading
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Configure plots
sns.set(style='whitegrid')

# Load dataset (ensure dataset.csv is in ../data/)
df = pd.read_csv('../data/dataset.csv')
print('Shape:', df.shape)
df.head(3)
```

```python
# Basic overview
display(df.head(10))
display(df.info())
display(df.describe(include='all'))
```

```python
# Missing values heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
plt.title('Missing Values Heatmap')
plt.show()
```

```python
# Distribution of Hazardous vs Non-Hazardous
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Hazardous')
plt.title('Hazardous vs Non-Hazardous Count')
plt.show()
```

```python
# Histograms for key numeric features
key_features = [
    'Relative Velocity km per hr',
    'Miles per hour',
    'Miss Dist.(Astronomical)',
    'Miss Dist.(lunar)',
    'Miss Dist.(kilometers)',
    'Semi Major Axis',
    'Aphelion Dist',
    'Mean Motion',
    'Orbital Period',
]
for col in key_features:
    if col in df.columns:
        plt.figure(figsize=(6, 4))
        sns.histplot(df[col].dropna(), kde=True)
        plt.title(f'Distribution: {col}')
        plt.show()
    else:
        print(f'[Info] Column not found: {col}')
```

```python
# Correlation heatmap (numeric-only)
plt.figure(figsize=(10, 8))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap (Numeric Features)')
plt.show()
```
